```python
model = tf.keras.models.load_model(root + 'classifier.h5')
```

```python
generator = tf.keras.models.load_model(root + 'GE W-InfoGAN Generator.h5')
```

```python
for i in range(c_cat):
    z = np.random.normal(size=(100,128))
    c = tf.keras.utils.to_categorical([i]*100, c_cat)
    c = c.reshape((-1,c_cat))
    image = generator.predict([z,c])
    image = image * 0.5 + 0.5
    # for i in range(10):
    #   plt.imshow(image[i,:,:,0], cmap='gray')
    #   plt.show()
    print(i , _names[i])
    y_ = model.predict(image)
    result =Counter(np.argmax(y_,axis=1))
    print('CNN',result.most_common())
    y_ = knn.predict(image.reshape((-1,128*128)))
    result =Counter(y_)
    print('knn',result.most_common())
```
